```python
def duration_time(value):        
    if ":" in value:
        game_duration = value.split(":")
        if len(game_duration) > 2:
            duration = (int(game_duration[0]) * 60) + int(game_duration[1]) + (int(game_duration[2]) / 60)
        else:
            duration = int(game_duration[0]) + (int(game_duration[1]) / 60)
        
        return duration
    
    else:
        return np.nan
    
```

```python
games_ovw_df["game_duration"] = games_ovw_df["game_duration"].astype(str).apply(duration_time).round(2)
games_ovw_df["game_map"] = [i.replace("-", "") for i in games_ovw_df["game_map"]]
games_ovw_df.head()
```

```python
query = """
WITH round_score AS (
    SELECT DISTINCT
        match_id,
        SUM (home_score) OVER (PARTITION BY match_id) AS home_total_round,
        SUM (away_score) OVER (PARTITION BY match_id) AS away_total_round
    FROM silver_fact_games
)
SELECT
    tour_id,
    m.match_id,
    date,
    bracket,
    patch_id,
    home_team_id,
    away_team_id,
    bo,
    home_score,
    away_score,
    ROUND ((1.0 * home_score - away_score) / (1.0 * home_score + away_score), 2) AS match_score_diff_pct,
    r.home_total_round AS home_total_round,
    r.away_total_round AS away_total_round,
    ROUND ((1.0 * home_total_round - away_total_round) / (1.0 * home_total_round + away_total_round), 2) AS match_round_diff_pct,
    home_h2h_win,
    away_h2h_win,
    ROUND ((1.0 * home_h2h_win - away_h2h_win) / (1.0 * home_h2h_win + away_h2h_win)) AS h2h_win_pct,
    home_h2h_score,
    away_h2h_score,
    ROUND ((1.0 * home_h2h_score - away_h2h_score) / (1.0 * home_h2h_score + away_h2h_score), 2) AS h2h_game_win_pct,
    home_n_last_win,
    away_n_last_win,
    home_n_last_match,
    away_n_last_match,
    away_n_last_wr,
    (CASE WHEN home_score > away_score THEN 1 ELSE 0 END) as is_home_win
FROM silver_fact_matches m
JOIN round_score r
ON r.match_id = m.match_id
"""

df = pd.read_sql(query, conn)
df.head()
```

```python
conn.execute("DROP VIEW IF EXISTS gold_view_team_map_statistics")

query = """
CREATE VIEW gold_view_team_map_statistics AS
WITH team_map_count AS (
    SELECT
        v.team_id,
        v.map_id,
        t1.total_matches,
        t2.total_win,

        SUM (CASE
            WHEN action = "pick"
            THEN 1
            ELSE 0
            END) AS pick_count,
        SUM (CASE
            WHEN action = "ban"
            THEN 1
            ELSE 0
            END) AS ban_count
    FROM gold_fact_map_vetos v
    JOIN (
    SELECT team_id, COUNT (DISTINCT match_id) AS total_matches
    FROM gold_view_team_game_statistics
    GROUP BY team_id
    ) t1
    ON v.team_id = t1.team_id
    JOIN (
    SELECT team_id, map_id, SUM (is_win) AS total_win
    FROM gold_view_team_game_statistics
    GROUP BY team_id, map_id
    ) t2
    ON v.team_id = t2.team_id
    AND v.map_id = t2.map_id
    GROUP BY v.team_id, v.map_id
),
team_map_rate AS (
    SELECT
        team_id,
        map_id,
        total_win,
        pick_count,
        ban_count,
        ROUND (1.0 * pick_count / total_matches, 2) AS pick_rate,
        ROUND (1.0 * ban_count / total_matches, 2) AS ban_rate,
        ROUND (1.0 * pick_count / (pick_count + ban_count), 2) AS map_pick_preference
    FROM team_map_count
)

SELECT
    s.team_id,
    s.map_id,
    COUNT (*) AS total_map_played,
    SUM (is_ot) AS total_ot_played,
    c.total_win,
    ROUND (1.0 * c.total_win / count (*), 2) as map_wr,
    c.pick_count,
    c.ban_count,
    c.pick_rate,
    c.ban_rate,
    c.map_pick_preference,
    ROUND (AVG (game_duration), 2) AS avg_game_duration,
    ROUND (AVG (round_diff_pct), 2) AS avg_round_diff_pct,
    ROUND (AVG (atk_wr), 2) AS avg_atk_wr,
    ROUND (AVG (def_wr), 2) AS avg_def_wr,
    ROUND (AVG (pstl_wr), 2) AS avg_pstl_wr,
    ROUND (AVG (eco_wr), 2) AS avg_eco_wr,
    ROUND (AVG (semi_eco_wr), 2) AS avg_semi_eco_wr,
    ROUND (AVG (semi_buy_wr), 2) AS avg_semi_buy_wr,
    ROUND (AVG (full_buy_wr), 2) AS avg_full_buy_wr
FROM gold_view_team_game_statistics s
JOIN team_map_rate c
ON s.team_id = c.team_id
AND s.map_id = c.map_id
GROUP BY s.team_id, s.map_id
;
"""

conn.execute(query)
conn.commit()

# df = pd.read_sql(query, conn)
# df
```

```python
conn.execute("DROP VIEW IF EXISTS gold_view_agent_statistics")

query = """
CREATE VIEW gold_view_agent_statistics AS
WITH game_count AS (
    SELECT
        COUNT(DISTINCT match_id) AS game_count
    FROM gold_fact_players
),
agent_presence_count AS (
    SELECT
        agent_id,
        COUNT (*) AS game_presence
    FROM (
        SELECT
            game_id,
            agent_id
        FROM gold_fact_players
        WHERE mod = "avg"
        GROUP BY game_id, agent_id
    )
    GROUP BY agent_id
)
SELECT
    p.agent_id,
    COUNT (*) AS pick_count,
    a.game_presence as game_presence_count,
    c.game_count,
    SUM (is_win) AS win_count,
    ROUND (1.0 * a.game_presence / c.game_count, 2) AS agent_presence_rate,
    ROUND (1.0 * COUNT(*) / c.game_count, 2) AS agent_pick_rate,
    ROUND (1.0 * SUM (is_win) / COUNT (*), 2) AS agent_win_rate
FROM gold_fact_players p
JOIN game_count c
JOIN agent_presence_count a
ON p.agent_id = a.agent_id
WHERE mod = "avg"
GROUP BY p.agent_id;
"""

conn.execute(query)
conn.commit()

# df = pd.read_sql(query, conn)
# df
```

```python
conn.execute("DROP VIEW IF EXISTS gold_view_agent_map_statistics")

query = """
CREATE VIEW gold_view_agent_map_statistics AS

WITH agent_map_count AS (
    SELECT
        agent_id,
        map_id,
        COUNT (*) AS agent_map_presence
    FROM (
        SELECT
            game_id,
            agent_id,
            map_id
        FROM gold_fact_players
        WHERE mod = "avg"
        GROUP BY game_id, agent_id, map_id
    )
    GROUP BY agent_id, map_id
)
SELECT
    p.agent_id,
    p.map_id,
    a.agent_map_presence AS agent_map_presence_count,
    COUNT (*) AS pick_count,
    SUM (is_win) AS win_count,
    ROUND (1.0 * SUM (is_win) / COUNT (*), 2) AS agent_win_rate,
    ROUND (1.0 * a.agent_map_presence / s.total_map_played, 2) AS agent_map_presence_rate
FROM gold_fact_players p
JOIN agent_map_count a
ON p.agent_id = a.agent_id
AND p.map_id = a.map_id
JOIN gold_view_map_statistics s
ON p.map_id = s.map_id
WHERE mod = "avg"
GROUP BY p.agent_id, p.map_id;
"""

conn.execute(query)
conn.commit()

# df = pd.read_sql(query, conn)
# df
```
